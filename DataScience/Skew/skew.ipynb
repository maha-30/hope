```python
# 1. Initialize DataFrame with all necessary indices
index_labels = [
    "Mean", "Median", "Mode", 
    "Q1:25%", "Q2:50%", "Q3:75%", "Q4:100%", 
    "IQR", "1.5Rule", "Lesser", "Greater", "Min", "Max"
]
descriptive = pd.DataFrame(index=index_labels, columns=quan)

# 2. Compute statistics for each column
for columnName in quan:
    # Safely convert column to numeric
    numeric_col = pd.to_numeric(dataset[columnName], errors='coerce')
    descriptive.loc["Mean", columnName] = numeric_col.mean()
    descriptive.loc["Median", columnName] = numeric_col.median()
    descriptive.loc["Mode", columnName] = numeric_col.mode()[0] if not numeric_col.mode().empty else None
   
    # Percentiles
    q1 = numeric_col.quantile(0.25)
    q2 = numeric_col.quantile(0.50)
    q3 = numeric_col.quantile(0.75)
    q4 = numeric_col.max()
    
    # Store basic values using .loc[row, column]
    descriptive.loc["Q1:25%", columnName] = q1
    descriptive.loc["Q2:50%", columnName] = q2
    descriptive.loc["Q3:75%", columnName] = q3
    descriptive.loc["Q4:100%", columnName] = q4
    
    # IQR and Outlier Calculations
    iqr = q3 - q1
    rule1_5 = 1.5 * iqr
    
    descriptive.loc["IQR", columnName] = iqr
    descriptive.loc["1.5Rule", columnName] = rule1_5
    descriptive.loc["Lesser", columnName] = q1 - rule1_5
    descriptive.loc["Greater", columnName] = q3 + rule1_5
    
    # Min and Max
    descriptive.loc["Min", columnName] = numeric_col.min()
    descriptive.loc["Max", columnName] = numeric_col.max()

print(descriptive)
```

```python
def freqTable(columnName,dataset):
    freqTable=pd.DataFrame(columns=["Unique_Values","Frequency","Relative_Frequency","Cumsum"])
    freqTable["Unique_Values"]=dataset[columnName].value_counts().index
    freqTable["Frequency"]=dataset[columnName].value_counts().values
    freqTable["Relative_Frequency"]=(freqTable["Frequency"]/103)
    freqTable["Cumsum"]=freqTable["Relative_Frequency"].cumsum()
    return freqTable
```
