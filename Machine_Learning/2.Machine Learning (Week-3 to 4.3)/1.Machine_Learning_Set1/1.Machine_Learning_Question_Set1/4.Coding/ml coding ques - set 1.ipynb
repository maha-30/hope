```python
#1.
Question 1:A retail store wants to identify customers who make frequent purchases. Given the dataset below, write a Python program to:
Group customers by their IDs.


Calculate the total purchase amount per customer.


Identify the top 3 customers with the highest purchase amounts.


Dataset:
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

Expected Output:
Total Purchases per Customer:
   Customer_ID  Purchase_Amount
0          101              520
1          102              600
2          103              480
3          104              300
4          105              400

Top 3 Frequent Customers:
   Customer_ID  Purchase_Amount
1          102              600
0          101              520
2          103              480
```

```python
import pandas as pd

# 1. Load dataset
data = {
    'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
    'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300],
}
df = pd.DataFrame(data)

# 2. Group by Customer_ID and calculate total purchase amount
total_purchases = (
    df.groupby('Customer_ID', as_index=False)['Purchase_Amount']
    .sum()
)

# 3. Identify the top 3 customers with the highest purchase amounts
top_3_customers = (
    total_purchases.sort_values(by='Purchase_Amount', ascending=False)
    .head(3)
)

# Display results
print("Total Purchases per Customer:")
print(total_purchases.to_string(index=True))

print("\nTop 3 Frequent Customers:")
print(top_3_customers.to_string(index=True))
```

```python
#2. Predicting House Prices with Linear Regression
A real estate company wants to predict house prices based on square footage. Write a Python program to:
Train a Linear Regression model.


Predict house prices for given test data.


Dataset:
data = {'Square_Feet': [1500, 2000, 2500, 3000, 3500],
        'Price': [300000, 400000, 500000, 600000, 700000]}

Test Data: [[1800], [2800]]
Expected Output:
Predicted Prices:
[360000. 560000.]

```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Prepare training dataset
data = {
    'Square_Feet': [1500, 2000, 2500, 3000, 3500],
    'Price': [300000, 400000, 500000, 600000, 700000],
}
df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[['Square_Feet']]
y = df['Price']

# 2. Train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# 3. Predict house prices for test data
test_data = [[1800], [2800]]
predicted_prices = model.predict(test_data)

# Display results
print("Predicted Prices:")
print(predicted_prices)
```

```python
#3:
Identifying Frequent Labels in a Dataset
A company wants to identify the top 3 most common categories in a dataset. Given the dataset below, write a Python program to:
Group the data by Category.


Count the total occurrences of each category.


Identify the top 3 most frequent categories.


Dataset:
data = {'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
        'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]}

Expected Output:

Total Occurrences per Category:
  Category  Count
0        A      4
1        B      3
2        C      3
3        D      1
4        E      1

Top 3 Frequent Categories:
  Category  Count
0        A      4
1        B      3
2        C      3
```

```python
import pandas as pd

# 1. Prepare dataset
data = {
    'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
    'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15],
}
df = pd.DataFrame(data)

# 2. Group by Category and count total occurrences
total_occurrences = (
    df.groupby('Category', as_index=False)
    .size()
    .rename(columns={'size': 'Count'})
)

# 3. Identify the top 3 most frequent categories
top_3_categories = (
    total_occurrences.sort_values(by='Count', ascending=False)
    .head(3)
)

# Display results
print("Total Occurrences per Category:")
print(total_occurrences.to_string(index=True))

print("\nTop 3 Frequent Categories:")
print(top_3_categories.to_string(index=True))
```

```python
#4.
Predicting Missing Values Using Mean Imputation
A dataset contains missing values in the Age column. Write a Python program to:
Replace missing values with the mean of the column.


Display the updated DataFrame.


Dataset:
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, None, 35, None]}

Expected Output:

Original Data:
     Name   Age
0  Alice  25.0
1    Bob  30.0
2 Charlie   NaN
3  David  35.0
4    Eve   NaN

Data after Imputation:
     Name   Age
0  Alice  25.0
1    Bob  30.0
2 Charlie  30.0
3  David  35.0
4    Eve  30.0
```

```python
import pandas as pd

# 1. Prepare dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, None, 35, None],
}
df = pd.DataFrame(data)

# Display original data
print("Original Data:")
print(df.to_string(index=True))

# 2. Impute missing values with the mean of the Age column
df['Age'] = df['Age'].fillna(df['Age'].mean())

# 3. Display updated DataFrame
print("\nData after Imputation:")
print(df.to_string(index=True))
```

```python
# 5: 
Implementing a Simple Linear Regression Model
You are given a dataset with Experience (years) and Salary ($). Write a Python program to:
Train a Linear Regression model.


Predict the salary for an individual with 6 years of experience.


Dataset:
import pandas as pd

data = {'Experience': [1, 2, 3, 4, 5], 
        'Salary': [30000, 35000, 40000, 45000, 50000]}
df = pd.DataFrame(data)

Expected Output (Example):

Predicted Salary for 6 years of experience: $55000


data = {'word_count': [100, 150, 200, 120, 180, 220], 
        'is_spam': ['ham', 'spam', 'spam', 'ham', 'spam', 'spam']}
df = pd.DataFrame(data)

Expected Output (Example):
Prediction for email with 200 words: Spam

```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Prepare dataset
data = {
    'Experience': [1, 2, 3, 4, 5],
    'Salary': [30000, 35000, 40000, 45000, 50000],
}
df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[['Experience']]
y = df['Salary']

# 2. Train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# 3. Predict salary for 6 years of experience
predicted_salary = model.predict([[6]])[0]

# Display result
print(f"Predicted Salary for 6 years of experience: ${int(predicted_salary)}")
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# 1. Prepare dataset
data = {
    'word_count': [100, 150, 200, 120, 180, 220],
    'is_spam': ['ham', 'spam', 'spam', 'ham', 'spam', 'spam'],
}
df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[['word_count']]
y = df['is_spam']

# 2. Train the Logistic Regression model
clf = LogisticRegression()
clf.fit(X, y)

# 3. Predict for an email with 200 words
pred = clf.predict([[200]])[0]

# Display result
print(f"Prediction for email with 200 words: {pred.capitalize()}")
```
